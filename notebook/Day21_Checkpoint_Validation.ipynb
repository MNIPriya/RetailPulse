```python

from pathlib import Path
import pandas as pd

PROJECT_DIR = Path(r"E:\RetailPulse")
OUTPUT_DIR = PROJECT_DIR / "outputs"
SCRIPT_DIR = PROJECT_DIR / "scripts"

# Files expected from previous project days
required_files = [
    "day2_cleaned_transactions.csv",
    "day2_sales_data.csv",
    "day2_rfm_features.csv",
    "day3_customer_segments.csv",
    "day4_daily_time_series.csv",
    "day5_prophet_metrics.csv",
    "day6_model_comparison.csv",
    "day8_ensemble_forecast.csv",
    "day9_churn_predictions.csv",
    "day10_inventory_recommendations.csv",
    "day12_drift_report.csv",
    "day13_daily_sales_summary.csv",
    "day14_validation_report.csv",
]

results = []

for filename in required_files:
    file_path = OUTPUT_DIR / filename
    results.append({
        "File": filename,
        "Exists": file_path.exists(),
        "Size_KB": (
            round(file_path.stat().st_size / 1024, 2)
            if file_path.exists()
            else None
        ),
    })

validation_df = pd.DataFrame(results)

display(validation_df)

missing_files = validation_df.loc[
    ~validation_df["Exists"], "File"
].tolist()

if missing_files:
    print("Files to investigate:")
    for filename in missing_files:
        print("-", filename)
else:
    print("All checked project output files are present.")

# Check the Streamlit dashboard
dashboard_path = SCRIPT_DIR / "streamlit_app.py"
print("\nDashboard exists:", dashboard_path.exists())

# Save checkpoint report
report_path = OUTPUT_DIR / "day21_validation_report.csv"
validation_df.to_csv(report_path, index=False)

print("Checkpoint report saved to:", report_path)
```
